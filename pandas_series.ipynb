```python
#create empty series

import pandas as pd
s=pd.Series()
print(s)
```

```python
#create a series from ndarray
#ndarray means-- n-dimensional array
#its 1-ndarray program
#amra ekhne index fixed kore dei ni tai by default 0 theke count korbe, count formula- range(len(array))-1

import pandas as pd
import numpy as np
data=np.array(['a','b','c','d'])
s=pd.Series(data)
print(s)
```

```python
#create a series from ndarray
#ndarray means-- n-dimensional array
#its 2-ndarray program
#amra ekhne index fixed kore dei ni tai by default 0 theke count korbe, count formula- range(len(array))-1

import pandas as pd
import numpy as np

# 2D-ndarray with string data
data = np.array([['a', 'b'], ['c', 'd']])  # Shape: (2, 2)

# Convert rows to list → then to Series
s = pd.Series(data.tolist())

print(s)
```

```python
#create a series from ndarray
#ndarray means-- n-dimensional array
#its 3-ndarray program
#amra ekhne index fixed kore dei ni tai by default 0 theke count korbe, count formula- range(len(array))-1

import pandas as pd
import numpy as np

#3D-ndarray with string data
data = np.array([['a', 'b', 'c'], ['d', 'e', 'f'], ['g', 'h', 'i']])  # Shape: (3, 3)

# Convert rows to list → then to Series
rows = data.tolist()
df = pd.Series(rows)

#df = pd.Series(data.tolist()) -- upore 2 line code(rows r df) ke ek sathe ei vabe lekha jay

print(df)



#its output show matrix from(3 x 3)
```

```python
#create a series from ndarray
#ndarray means-- n-dimensional array
#its 3-ndarray program
#amra ekhne index fixed kore dibo tai by default 0 theke count korbe nh, count formula- range(len(array))-1

import pandas as pd
import numpy as np

#3D-ndarray with string data
data = np.array([['a', 'b', 'c'], ['d', 'e', 'f'], ['g', 'h', 'i']])  # Shape: (3, 3)

df = pd.Series((data.tolist()), index = [100,101,102])
print(df)



#its output show matrix from(3 x 3)
#jodi index fixed korte chai taile ei vbe krbo
```

**Create a series from dictionary**

---



```python
# create a series from dictionary
#Duplicate keys in dictionary not allowed

# Dictionary theke Series banate hole, keys gulo automatically index hoye jabe, ar values gulo Series er values hobe.
#key = {a, b, c, d, e}
#values = {10, 20, 30, 40, 50}

import pandas as pd
import numpy as np

data = {'a':10, 'b':20, 'c':30, 'd':40, 'e':50}

s = pd.Series(data)
print(s)
```

```python
#Duplicate keys in dictionary not allowed
#ami kono dict te nijer icche mto index bosaite pari but jodi index er key dict te nh thake tkhn seta nan ashbe
#Custom index, Allowed, order can change

#pd.Series(dictionary, index=[custom_index_list])


import pandas as pd
import numpy as np

data = {'a':1.23, 'b':3.33, 'c':5.0, 'd':1, 'e':10}

s = pd.Series((data), index = ['b', 'c', 'd', 'a', 'e', 'f'])
s
```

```python
#  Series supports values as list, tuple, string, etc.
#  dtype hobe object if mixed types are used.

data = {'a': [1, 2], 'b': (3, 4), 'c': 'hello'}
s = pd.Series(data)
print(s)
```

***create s series from scalar***

---



What is a scalar?

A scalar mane hocche ekta single value. example:-- 5, 'hello', True, 3.14

Eta kono list, tuple, dict, ba array na — just one value.

r akta value thkle amra bolbo scalar series.

ok?

```python
import pandas as pd

# Create Series from scalar
s = pd.Series(5, index=['a', 'b', 'c', 'd', 'e'])
print(s)
```

```python
import pandas as pd

# Create Series from scalar list
s = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])

#print middle number
print(s[1:4])

# Print last 3 elements
print(s[-3:])

# Print last 2 elements
print(s[3:])
```

**Retrieve data using label(index)**

---



```python
# Single label access

import pandas as pd

s = pd.Series([1,2,3,4,5], index = ['a', 'b', 'c', 'd', 'e'])

#retrieve a single element
print(s['a'])
```

```python
# Multiple labels (list of labels)

import pandas as pd

s = pd.Series([1,2,3,4,5], index = ['a', 'b', 'c', 'd', 'e'])

#retrieve multiple elements
print(s[['a', 'c', 'd']])
```

```python
# Label slicing (if labels are sorted)


import pandas as pd


s = pd.Series([1,2,3,4,5], index = ['a', 'b', 'c', 'd', 'e'])


print(s['a':'b'])  # Includes both 'a' and 'b'

print() # create new line

print(s['a':'e'])  # Includes both 'a' and 'e'
```

```python
# amr index ba series theke ki vabe element er position ber korbo tar method
# ei method amr (create s series from scalar) koreci


import pandas as pd
s = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])

#print first series.number
print(s[0])

print("-------------------------------------")

#print middle s.number
print(s[1:4])

print("-------------------------------------")

# Print last 3 s.number
print(s[-3:])

print("-------------------------------------")

# Print last 2 s.number
print(s[3:])
```
